### Monte Carlo Every Visit Prediction

Monte Carlo Every Visit Prediction is a method in reinforcement learning used to estimate the value function for a given policy. In 'every-visit' MC prediction, every time a state is visited in an episode, its return is accumulated, and the average is taken over all such visits. This approach is suitable for episodic tasks where the agent interacts with the environment, and the episode eventually terminates.

```python
import numpy as np
import random
from collections import defaultdict
```

```python
class SimpleEnvironment:
    def __init__(self):
        self.states = list(range(12, 22))
        self.actions = [0, 1]
    def step(self, state, action):
        if action == 0:
            new_card = random.randint(1, 10)
            new_state = state + new_card
            if new_state > 21:
                return new_state, -1, True
            elif new_state == 21:
                return new_state, 1, True
            else:
                return new_state, 0, False
        else:
            dealer_sum = random.randint(15, 21)
            if state > dealer_sum:
                return state, 1, True
            elif state < dealer_sum:
                return state, -1, True
            else:
                return state, 0, True
    def reset(self):
        return random.randint(12, 18)

def generate_episode(env, policy):
    episode = []
    state = env.reset()
    done = False
    while not done:
        action = policy(state)
        next_state, reward, done = env.step(state, action)
        episode.append((state, action, reward))
        state = next_state
    return episode
def simple_policy(state):
    if state < 17:
        return 0
    else:
        return 1
```

```python
def monte_carlo_every_visit_prediction(env, policy, num_episodes):
    returns = defaultdict(list)
    V = defaultdict(float)
    for i_episode in range(1, num_episodes + 1):
        episode = generate_episode(env, policy)
        states = [s for s, a, r in episode]
        rewards = [r for s, a, r in episode]
        G = 0
        for t in range(len(episode) - 1, -1, -1):
            R_t = rewards[t]
            S_t = states[t]
            G = R_t + G
            returns[S_t].append(G)
            V[S_t] = np.mean(returns[S_t])
    return V
```

```python
env = SimpleEnvironment()
num_episodes = 10000
V_every_visit = monte_carlo_every_visit_prediction(env, simple_policy, num_episodes)
print("Estimated State-Value Function (Every-Visit MC):")
for state in sorted(V_every_visit.keys()):
    print(f"V({state}): {V_every_visit[state]:.4f}")
```
