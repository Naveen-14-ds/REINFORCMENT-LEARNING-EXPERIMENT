```python
import gym
import numpy as np
np.bool8 = np.bool_
env = gym.make('CartPole-v0')
state = env.reset()
print(f"Action Space: {env.action_space}")
print(f"Observation Space: {env.observation_space}")
def describe_state(state):
    print(f"Cart Position: {state[0]}")
    print(f"Cart Velocity: {state[1]}")
    print(f"Pole Angle: {state[2]}")
    print(f"Pole Angular Velocity: {state[3]}")
print("Initial State:")
describe_state(state)
actions = {0: "Move Left", 1: "Move Right"}
for action in actions:
    print(f"Action {action}: {actions[action]}")
num_steps = 5
print("\nSimulating a few steps:")
for step in range(num_steps):
    action = env.action_space.sample()
    next_state, reward, done, info = env.step(action)
    print(f"\nStep {step + 1}:")
    print(f"Action taken: {actions[action]}")
    print("Next State:")
    describe_state(next_state)
    print(f"Reward: {reward}")
    print(f"Done: {done}")
env.close()
```

```python
import gym
print("\n--- Exploring FrozenLake-v1 Environment ---")
env = gym.make('FrozenLake-v1', is_slippery=False)
print(f"Action Space: {env.action_space}")
print(f"Observation Space: {env.observation_space}")
print("\n--- MDP Components of FrozenLake-v1 ---")
print("**States:**")
print("The environment is a 4x4 grid. Each cell represents a state. There are 16 states (0-15). S = Start, F = Frozen, H = Hole, G = Goal.")
print("For example, in a 4x4 grid:")
print("0  1  2  3")
print("4  5  6  7")
print("8  9 10 11")
print("12 13 14 15")
print("\n**Actions:**")
print("The agent can take 4 discrete actions:")
print("0: Left")
print("1: Down")
print("2: Right")
print("3: Up")
print("\n**Transition Probabilities (P):**")
print("This is a dictionary that maps (state, action) pairs to a list of possible outcomes. Each outcome is a tuple: (probability, next_state, reward, done). For `is_slippery=False`, probability is 1.0.")
print("Example for state 0 (Start) and action 1 (Down):")
print(f"P[0][1] (from state 0, action Down): {env.P[0][1]}")
print("This means from state 0, taking action Down leads to state 4 with probability 1.0, reward 0.0, and the episode is not done.")
print("\n**Reward:**")
print("The reward structure is as follows:")
print("- +1 reward for reaching the Goal (G).")
print("- 0 reward for reaching a Hole (H).")
print("- 0 reward for all other transitions (Frozen or Start cells).")
print("The episode terminates when the agent reaches a Goal (G) or falls into a Hole (H).")
env.close()
```
